```python
# Cell 1: Definisi Fungsi f(x) dan Implementasi Metode Bisection
def f(x):
    # Contoh fungsi: f(x) = x^3 - x - 2
    return x**3 - x - 2

def bisection_method(a, b, tol=1e-6, max_iter=100):
    # Validasi awal apakah interval mengapit akar
    if f(a) * f(b) >= 0:
        raise ValueError("Fungsi harus memiliki tanda yang berlawanan pada titik a dan b.")

    c = a
    print(f"{'Iterasi':<8} {'a':<10} {'b':<10} {'c (Tengah)':<12} {'f(c)':<12}")
    print("-" * 55)

    for i in range(max_iter):
        c = (a + b) / 2.0
        fc = f(c)

        print(f"{i+1:<8} {a:<10.5f} {b:<10.5f} {c:<12.5f} {fc:<12.5f}")

        # Cek kriteria konvergensi (toleransi atau akar tepat)
        if abs(fc) < 1e-12 or (b - a) / 2.0 < tol:
            break

        # Menentukan interval selanjutnya
        if f(a) * fc < 0:
            b = c
        else:
            a = c

    return c
```

```python
# Cell 2: Menjalankan metode bisection pada interval [1, 2]
a_batas = 1.0
b_batas = 2.0
akar_hasil = bisection_method(a_batas, b_batas)
```

```python
# Cell 3: Menampilkan Hasil Akhir Akar Bisection
print("\n=== HASIL AKHIR METODE BISECTION ===")
print(f"Akar hampiran persamaan ditemukan pada x = {akar_hasil:.6f}")
```
